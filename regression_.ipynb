```python
# Step 1 - Creating a dataframe

import pandas as pd

# Upload your file to Google Colab
from google.colab import files
uploaded = files.upload()

# Replace 'your_file.xlsx' with the actual name of your Excel file
excel_file = 'Task3.xlsx'

# Read the Excel file into a DataFrame
df = pd.read_excel(excel_file)

# Display the DataFrame
print(df)
```

```python
# Step 2 - Display Summary Statistics

summary_stats = df.describe()
print(summary_stats)
```

```python
# Step 3 - Running a regression model

import numpy as np
import statsmodels.api as sm

# Add a constant term for the intercept. Define independent and dependent variables

X = sm.add_constant(df[['SQFT', 'BEDS', 'BATHS', 'CARS', 'AGE']])
Y = df['Price']

# Fit the model
model = sm.OLS(Y, X).fit()

# Display the summary, including coefficients and p-values
print(model.summary())
```
